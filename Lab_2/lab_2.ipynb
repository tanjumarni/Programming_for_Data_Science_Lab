```python
#problem 1
marks = [95, 73, 80, 90, 70]

total = 0
passed = 0
highest = marks[0]
lowest = marks[0]

for mark in marks:
    total += mark

    if mark >= 50:
        passed += 1

    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

average = total / len(marks)

print("Marks:", marks)
print("Total:", total)
print("Average:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Passed Courses:", passed)
```

```python
#problem 2
temperatures = (32, 30, 25, 37, 41, 28)

for i, temperature in enumerate(temperatures):
    if i < 3:
        day = "Saturday"
    else:
        day = "Sunday"

    print(day, "Temperature:", temperature)

print("Highest Temperature:", max(temperatures))
print("Lowest Temperature:", min(temperatures))
```

```python
#problem 3
text = input("Enter a word or short sentence: ")

vowels = 0
spaces = 0

for character in text:
    if character.lower() in "aeiou":
        vowels += 1

    if character == " ":
        spaces += 1

print("Characters:", len(text))
print("Vowels:", vowels)
print("Spaces:", spaces)
```

```python
#problem 4
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


marks = [85, 75, 65, 80, 70]

print("Average:", sum(marks) / len(marks))
print("Grade:", calculate_grade(marks))

```

```python
#problem 5
products = [
    {"name": "Ice Cream", "price": 600, "quantity":2},
    {"name": "Chips", "price": 50, "quantity": 15},
    {"name": "Milk", "price": 200, "quantity": 3},
    {"name": "cake", "price": 1000, "quantity": 1}
]

total = 0

for product in products:
    total += product["price"] * product["quantity"]


def apply_discount(total):
    if total >= 2000:
        return total * 0.10
    else:
        return total


print("Total Cost:", total)
print("Final Cost:", apply_discount(total))
```

```python
#Problem 7
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total = 0
passed = 0
highest = students[0]

for student in students:
    print("Name:", student["name"], "Marks:", student["marks"])

    total += student["marks"]

    if student["marks"] >= 50:
        passed += 1

    if student["marks"] > highest["marks"]:
        highest = student

average = total / len(students)

print("Passed Students:", passed)
print("Class Average:", average)
print("Highest Mark:", highest["marks"])
print("Top Student:", highest["name"])
```

```python
#Problem 8
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

total = 0
highest = 0
highest_product = ""

for product in products:
    line_total = product["price"] * product["quantity"]

    print(product["name"], "Line Total:", line_total)

    total += line_total

    if line_total > highest:
        highest = line_total
        highest_product = product["name"]

print("Total Inventory Value:", total)
print("Highest Line Total:", highest)
print("Highest Value Product:", highest_product)
```
