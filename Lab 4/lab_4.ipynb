```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


df = pd.read_csv("train.csv")

print(df.head())
```

```python
#Problem 1
print(df.info())

print(df.select_dtypes(include=['int64','float64']).columns)

print(df.select_dtypes(include=['object']).columns)
```

```python
#problem 2

average_fare = df.groupby('Pclass')['Fare'].mean()

plt.plot(average_fare.index, average_fare.values)


plt.title("Average Fare Across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")

plt.show()
```

```python
#problem 3

plt.scatter(df['Age'], df['Fare'])

plt.title("Relationship Between Passenger Age and Fare")
plt.xlabel("Passenger Age")
plt.ylabel("Fare")


plt.show()
```

```python
#problem 4
gender_count=df['Sex'].value_counts()

plt.bar(gender_count.index,gender_count.values)

plt.title("Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
#problem 5
plt.hist(df['Age'].dropna(), bins=10)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
#problem 6

survival_count = df['Survived'].value_counts()

plt.pie(
    survival_count,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%"
)


plt.title("Proportion of Passengers Who Survived")

plt.show()
```

```python
#problem 7

fig, ax = plt.subplots(1, 2)


ax[0].hist(df['Age'].dropna(), bins=10)
ax[0].set_title("Distribution of Passenger Ages")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Frequency")


class_count = df['Pclass'].value_counts().sort_index()

ax[1].bar(class_count.index, class_count.values)
ax[1].set_title("Number of Passengers in Each Class")
ax[1].set_xlabel("Passenger Class")
ax[1].set_ylabel("Number of Passengers")


plt.tight_layout()

plt.show()
```

```python
#problem 8

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

gender_count = df['Sex'].value_counts()

ax[0].bar(gender_count.index, gender_count.values)
ax[0].set_title("Passengers by Gender")
ax[0].set_xlabel("Gender")
ax[0].set_ylabel("Number of Passengers")

ax[1].hist(df['Fare'], bins=10)
ax[1].set_title("Distribution of Passenger Fares")
ax[1].set_xlabel("Fare")
ax[1].set_ylabel("Frequency")


survival_count = df['Survived'].value_counts()

ax[2].pie(
    survival_count,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%"
)

ax[2].set_title("Passenger Survival Proportion")


plt.tight_layout()

plt.show()
```
