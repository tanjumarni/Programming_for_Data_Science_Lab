```python
#problem 1
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("Sum:", num1 + num2)
print("Difference:", num1 - num2)
print("Product:", num1 * num2)

if num2 != 0:
    print("Quotient:", num1 / num2)
else:
    print("Quotient:Division by zero is not possible ")
```

```python
#problem 3
celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

print("Celsius: ", celsius, "°C")
print("Fahrenheit: ", fahrenheit, "°F")
print("Kelvin: ", kelvin, "K")
```

```python
#problem 4
name = input("Enter student name: ")

English = float(input("Enter English marks: "))
Accounting = float(input("Enter Accounting marks: "))
Software_Engineering = float(input("Enter Software Engineering marks: "))
python  = float(input("Enter Python marks: "))
Economices = float(input("Enter Economices marks: "))

total = 0
passed = 0
highest = 0
lowest = 100

for mark in [English,Accounting ,Software_Engineering,python,Economices ]:
    total += mark

    if mark >= 50:
        passed += 1

    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

average = total / 5

print()
print("Student Name:", name)
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Passed Courses:", passed)
```
