```python
#Selecting and Filtering Data

# Select one column & multiple columns
print(df["Name"].head())
df[["Name", "Age", "Sex", "Survived"]].head()
```

```python
#Select rows using labels
print(df.loc[0:4, ["Name", "Age"]])

#Select rows using integer positions
print(df.iloc[0:5, 0:4])
```

```python
# Passengers older than 30
print(df[df["Age"] > 30])
# Female passengers
print(df[df["Sex"] == "female"])
# Passengers who survived
print(df[df["Survived"] == 1])
# Female passengers who survived
print(df[
(df["Sex"] == "female") &
(df["Survived"] == 1)])
```

```python
#Problem-2
# 1. First-class passengers
first_class = df[df["Pclass"] == 1]
print("First-class passengers:", len(first_class))
first_class.head()
```

Keeps only rows where Pclass equals 1.len() counts the rows returned.

```python
# 2. Passengers younger than 18
under_18 = df[df["Age"] < 18]
print("Passengers younger than 18:", len(under_18))
under_18.head()
```

Rows with missing Age are automatically excluded because comparisons with missing values are False.

```python
# 3. Male passengers who survived
male_survived = df[(df["Sex"] == "male") & (df["Survived"] == 1)]
print("Male passengers who survived:", len(male_survived))
male_survived.head()
```

Two conditions are combined with &, so a row must be male and survived.

```python
# 4. Passengers who paid a fare greater than 100
high_fare = df[df["Fare"] > 100]
print("Passengers with fare > 100:", len(high_fare))
high_fare.head()
```

Filters passengers whose Fare is above 100.

```python
# Sorting and Aggregation
# Sort passengers by age
print(df.sort_values(by="Age"))

# Sort by fare in descending order
print(df.sort_values(
     by="Fare",
     ascending=False
 ))
# Sort by passenger class and fare
print(df.sort_values(
   by=["Pclass", "Fare"]
))
```

```python
#Basic Aggregation
# Average passenger age
print(df["Age"].mean())

# Median passenger fare
print(df["Fare"].median())

# Maximum and minimum fare
print(df["Fare"].max())
print(df["Fare"].min())

# Count passengers by gender
print(df["Sex"].value_counts())
# Average fare by passenger class
print(df.groupby("Pclass")["Fare"].mean())

# Survival rate by gender
print(df.groupby("Sex")["Survived"].mean())
```

Final Laboratory Challenge

```python
# Q1.How many passengers are in the dataset?
total_passengers = len(df)
print("Total passengers:", total_passengers)
```

Answer 1:There are 891 passengers in the dataset.

```python
# Q2.What is the average passenger age?
print("Average age:", round(df["Age"].mean(), 2))
```

Answer 2:The average passenger age is about 29.70 years (missing ages are ignored).

```python
# Q3.Which passenger class has the highest number of passengers?
class_counts = df["Pclass"].value_counts()
print(class_counts)
print("Class with most passengers:", class_counts.idxmax())
```

Answer 3:Class 3 has the most passengers (491), compared with 216 in class 1 and 184 in class 2.

```python
# Q4.What is the average fare paid by male and female passengers?
print(df.groupby("Sex")["Fare"].mean().round(2))
```

Answer 4:Female passengers paid an average of about 44.48, while male passengers paid about 25.52. Women paid more on average, probably because a larger share of the women travelled in first class.

```python
# Q5.Which gender has a higher survival rate?
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100
print(survival_by_sex.round(2))
print("Higher survival rate:", survival_by_sex.idxmax())
```

Answer 5:Females have a much higher survival rate (74.20%) than males (18.89%).

```python
# Q6.How does the survival rate differ across passenger classes?
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100
print(survival_by_class.round(2))
```

Answer 6:Survival falls as class gets lower:62.96% in first class, 47.28% in second class and 24.24% in third class.
